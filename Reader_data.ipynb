```python
import csv

def read_generation_data(filename, start_gen, end_gen):
    generation_data = []
    
    with open(filename, mode='r') as file:
        reader = csv.reader(file)
        next(reader)  # Salta l'intestazione
        for row in reader:
            gen = int(row[0])  # La colonna 'Generation'
            if start_gen <= gen <= end_gen:
                individual_data = {
                    'Generation': gen,
                    'Individual': int(row[1]),
                    'Genome': row[2],
                    'Score': float(row[3])
                }
                generation_data.append(individual_data)
    
    return generation_data
```

```python
start_gen = 0
end_gen = 2
data = read_generation_data('genome_score_per_generation.csv', start_gen, end_gen)
print(data)
```
